```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('../classification_results_trial_0001.csv')
data.head()
```

1. Quantas imagens são “benign” e “malign” em real_class? 


```python
data['real_class'].value_counts()
```

2. Identifique em quais imagens o modelo errou a predição. 


```python
erros = data[data['real_class']!=data['predicted_class']]
erros['image_path']
```

3. Verifique se o modelo estava confiante mesmo quando errou. 


```python
# Ver a probabilidade da classe predita (confiança)
erros['confianca'] = erros.apply(
    lambda row: row['prob_malign'] if row['predicted_class'] == 'maligna' else row['prob_benign'],
    axis=1
)

erros['confianca'].describe()
```

```python
erros['confianca']
```

4. Calcule: <br>
a. TP (real malign, previsto malign) <br>
b. TN (real benign, previsto benign) <br>
c. FP (real benign, previsto malign) <br>
d. FN (real malign, previsto benign) <br>

```python
TP = len(data[(data['real_class'] == 'malign') & (data['predicted_class'] == 'malign')])
TN = len(data[(data['real_class'] == 'benign') & (data['predicted_class'] == 'benign')])
FP = len(data[(data['real_class'] == 'benign') & (data['predicted_class'] == 'malign')])
FN = len(data[(data['real_class'] == 'malign') & (data['predicted_class'] == 'benign')])
```

```python
print(f'TP: {TP}, TN: {TN}, FP: {FP}, FN: {FN}')
```

5. Calcule: <br>
a. Acurácia: (TP+TN)/(TP+TN+FP+FN) <br>
b. Precisão: TP/(TP+FP) <br>
c. Recall: TP/(TP+FN) <br>
d. Especificidade: TN/(TN+FP) <br>

```python
accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)
specificity = TN / (TN + FP)

print(f'Accuracy: {accuracy:.2f}')
print(f'Precision: {precision:.2f}')
print(f'Recall: {recall:.2f}')
print(f'Specificity: {specificity:.2f}')
```
